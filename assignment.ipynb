```python
# Patients table

patients_data = {
    'patient_id': [101, 102, 103, 104, 105],
    'name': ['Alice Chen', 'Bob Wilson', 'Carol Davis', 'David Kim', 'Eva Martinez'],
    'age': [32, 45, 28, 60, 38],
    'gender': ['F', 'M', 'F', 'M', 'F'],
    'blood_type': ['A+', 'O-', 'B+', 'AB+', 'A-'],
    'admission_date': ['2023-01-15', '2023-02-20', '2023-03-10', '2023-01-05', '2023-04-18']
}

# Doctors table

doctors_data = {
    'doctor_id': [1, 2, 3],
    'name': ['Dr. Smith', 'Dr. Johnson', 'Dr. Lee'],
    'specialization': ['Cardiology', 'Neurology', 'Pediatrics'],
    'years_experience': [10, 8, 5]
}

# Appointments table

appointments_data = {
    'appointment_id': [1001, 1002, 1003, 1004],
    'patient_id': [101, 102, 103, 105],
    'doctor_id': [1, 2, 3, 1],
    'date': ['2023-05-10', '2023-05-12', '2023-05-15', '2023-05-18'],
    'diagnosis': ['Hypertension', 'Migraine', 'Common Cold', 'Arrhythmia']
}

# Medications table

medications_data = {
    'medication_id': [501, 502, 503],
    'name': ['Lisinopril', 'Sumatriptan', 'Amoxicillin'],
    'category': ['Blood Pressure', 'Migraine', 'Antibiotic'],
    'cost': [20.50, 35.75, 8.99]
}

# Prescription table

prescriptions_data = {
    'prescription_id': [201, 202, 203],
    'appointment_id': [1001, 1002, 1004],
    'medication_id': [501, 502, 501],
    'dosage': ['10mg daily', '50mg as needed', '5mg twice daily']
}

patients_data_df = pd.DataFrame(patients_data)
doctors_data_df = pd.DataFrame(doctors_data)
appointments_data_df = pd.DataFrame(appointments_data)
medications_data_df = pd.DataFrame(medications_data)
prescriptions_data_df = pd.DataFrame(prescriptions_data)    
```

```python

inner_join_appointments = pd.merge(
    appointments_data_df, 
    patients_data_df, 
    on='patient_id', 
    how='inner'
)
 
```

```python
inner_join_appointments
```

```python
outer_join_medications = pd.merge(
    medications_data_df,
    prescriptions_data_df, 
    on='medication_id',             
    how='outer'
)
```

```python
# Students Table
students = pd.DataFrame({
    'student_id': [1, 2, 3, 4, 5],
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'class': ['10A', '10B', '10A', '10C', '10B']
})

# Scores Table
scores = pd.DataFrame({
    'student_id': [3, 4, 5, 6],
    'math_score': [88, 92, 75, 85],
    'science_score': [90, 85, 80, 70]
})

students_data_df = pd.DataFrame(students)
scores_data_df = pd.DataFrame(scores)
```

```python
inner_join_students = pd.merge(students_data_df,  scores_data_df,  on='student_id',  how='inner')
inner_join_students
```

```python
outer_join_students = pd.merge(students_data_df,  scores_data_df,  on='student_id',  how='outer')
outer_join_students
```

```python
left_join_students = pd.merge(students_data_df,  scores_data_df,  on='student_id',  how='left')
left_join_students
```

```python
right_join_students = pd.merge(students_data_df,  scores_data_df,  on='student_id',  how='right')
right_join_students
```

```python
left_join_students_df = pd.merge(students_data_df, scores_data_df, on='student_id', how='left').fillna(0)
left_join_students_df
```

```python
df = pd.merge(students, scores, on='student_id', how='left').fillna(0)
df['total_score'] = df['math_score'] + df['science_score']
print(df)
```
