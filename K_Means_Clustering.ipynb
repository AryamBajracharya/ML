```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
from sklearn.datasets import make_blobs  # Random data points bata halka cluster represent garxa tara acutally ma cluster haina
X, y_true = make_blobs(n_samples=300, centers=3, cluster_std=0.60, random_state=0)
plt.scatter(X[:,0], X[:,1], s = 50)
plt.show()
```

```python
# K_MEANS_ALGORITHM

kmeans = KMeans(n_clusters=3, random_state=0)
kmeans.fit(X)
y_kmeans = kmeans.predict(X)

plt.scatter(X[:,0], X[:,1], c=y_kmeans, s=50, cmap='viridis')
centres = kmeans.cluster_centers_
plt.scatter(centres[:,0], centres[:,1], c='red', s=200, alpha=0.75, marker='X')
plt.show()
```

```python
# ELBOW Method

X, _ = make_blobs(n_samples=300, centers=5, cluster_std=0.60, random_state=0)
plt.scatter(X[:,0], X[:,1], s = 50)
plt.show()
```

```python
# Calculate inertia

inertia = []
k_values = range(1,10)

for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=0)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)
```

```python
plt.plot(k_values, inertia, marker = 'o')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow method for optimal k')
plt.show()
```

```python
from kneed import KneeLocator
knee = KneeLocator(k_values, inertia, curve='convex', direction='decreasing')
print(f'Optimal number of clusters according to Elbow Method: {knee.knee}')
```

```python
plt.plot(k_values, inertia, marker ='o')
plt.vlines(knee.knee, plt.ylim()[0], plt.ylim()[1], linestyles='dotted', color='salmon')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method with Automated Knee Detection')
plt.show()
```
