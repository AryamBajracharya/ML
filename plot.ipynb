```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
data = {
    'Year': [2010, 2011, 2012, 2013, 2014, 2015, 2016],
    'Sales': [100, 150, 200, 250, 300, 350, 400],
    'Profit': [20, 30, 50, 70, 90, 100, 110]
}

df = pd.DataFrame(data)
df
```

```python
# BAR PLOT

# Using Matplotlib

plt.figure (figsize=(10, 5))   # Kati size ko figure chainxa vanera
plt.bar(df['Year'], df['Sales'],color='skyblue')
plt.xlabel('Year')
plt.ylabel('Sales')
plt.title('Sales Over Years')
plt.show()
```

```python
# Using Seaborn

plt.figure(figsize=(10, 5))
sns.barplot(x='Year', y='Sales',data=df)
plt.xlabel('Year')
plt.ylabel('Sales') 
plt.title('Sales Over Years')
plt.show()
```

```python
# LINE PLOT

# Using Matplotlib

plt.figure(figsize=(10, 5))
plt.plot(df['Year'], df['Sales'], marker='*', color='red', linestyle ='--')
plt.xlabel('Year')
plt.ylabel('Sales')
plt.title('Sales Over Years')
plt.show()
```

```python
# Using Seaborn
plt.figure(figsize=(10, 5))
sns.lineplot(x='Year', y='Sales',data=df, marker='<', color='green')
plt.title('Sales Over Year')
plt.grid(True)  # Background ma lines haru aauxa
```

```python
# Histogram

# Using matplotlib

import numpy as np

data = np.random.normal(loc=50, scale=10, size=200)
df = pd.DataFrame({'Score':data})
df
```

```python
# Plotting in Histogram

plt.figure(figsize=(10, 5))
plt.hist(df['Score'], bins=20, edgecolor='black', color='red') #Bins = kati area cover garne vanera hunxa in:count 20
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.title('Distribution of Data')
plt.grid(True)
plt.show()
```

```python
# Using Seaborn

sns.histplot(x='Score', data=df, bins=20, kde=True, color='salmon')
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

```python
# BOX PLOT

math_scores = [65,79,75,80,85,90,95,100,150]
english_scores = [55,60,65,70,75,80,85,90,95]

plt.figure(figsize=(10,5))
plt.boxplot([math_scores, english_scores],labels=['Math','English'])
plt.title('Boxplot of Math and English Score')
plt.xlabel('Subject')
plt.ylabel('Scores')
plt.grid(True)
plt.show()
```

```python
# Using Seaborn

df = pd.DataFrame({
    'Subject': ['Math']*7 + ['English']*7,
    'Scores': [89,92,80,89,100,76,96,86,92,30,40,60,70,80]
    })

sns.boxplot(x='Subject', y='Scores', data=df)
plt.title('Seaborn Box Plot')
plt.show()
```

```python
# HEAT-MAP

data = np.array([[30,20,40],
                [25,35,30],
                [20,25,45]
])
df= pd.DataFrame(data, columns=['Math','Science','English'],
                 index=['class_a','class_b','class_c'])
df
```

```python
# Using matplotlib

plt.imshow(data, cmap='YlGnBu')
plt.title('Heatmap')
plt.xticks([0,1,2], df.columns.tolist())
plt.yticks([0,1,2], df.index.tolist())
plt.colorbar(label='Scores')
plt.show()
```

```python
# Using Seaborn

sns.heatmap(df, annot=True, cmap='YlGnBu') # ANNOT = TURE GARDA BHITRA KO VALUE NI AAUXA    
```
