```python
import numpy as np
```

```python
# Sigmoid and its derivative
def sigmoid(x):
    return 1/(1+np.exp(-x))

def sigmoid_deriv(x):
    sx = sigmoid(x)
    return sx* (1-sx)
```

```python
# Input and Target
X = np.array([[0.5, 0.9]])
y_true = np.array([[1]])
```

```python
print(f'The shape of X is{X.shape}')
print(f'The shape of y is {y_true.shape}')
```

```python
# Weights and Biases
np.random.seed(42)
w1 = np.random.rand(2,2)
b1 = np.zeros((1,2))

w2 = np.random.rand(2,1)
b2 = np.zeros((1,1))
```

```python
w1.shape
```

```python
b2.shape
```

```python
# Training Settings
lr = 0.1
epochs = 1000
```

```python
for epoch in range(epochs):
    # Forward Propagation 
    z1 = np.dot(X, w1) + b1
    a1 = sigmoid(z1)

    z2 = np.dot(a1, w2) + b2
    y_pred= sigmoid(z2)

    # Loss Calculation
    loss = np.mean((y_true - y_pred) ** 2)

    # Backward Propagation
    d_loss_y_pred = 2 * (y_pred-y_true)
    d_y_pred_z2 = sigmoid_deriv(z2)
    d_z2_w2 = a1.T

    d_loss_w2 = np.dot(d_z2_w2, d_loss_y_pred * d_y_pred_z2)
    d_loss_b2 = d_loss_y_pred * d_y_pred_z2

    d_z2_a1 = w2
    d_a1_z1 = sigmoid_deriv(z1)
    d_z1_w1 = X.T

    d_loss_w1 = np.dot(d_z1_w1, (d_loss_y_pred * d_y_pred_z2 @ d_z2_a1.T) * d_a1_z1)
    d_loss_b1 = (d_loss_y_pred * d_y_pred_z2 @ d_z2_a1.T) * d_a1_z1

    # Update Weights and Biases
    w2 -= lr * d_loss_w2
    b2 -= lr * d_loss_b2

    w1 -= lr * d_loss_w1
    b1 -= lr * d_loss_b1

    if epoch % 100==0:
        print(f'Epoch {epoch}')
        print(f'Acutal (y_true): {y_true[0][0]:.4F}')
        print(f'Predicted (y_pred): {y_pred[0][0]:.4F}')
        print(f'Loss: {loss:.6F}\n')
```
