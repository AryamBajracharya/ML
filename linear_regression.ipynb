```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-1,1,  100).reshape(-1, 1)
```

```python
x
```

```python
y = 2 * x + 1 + 0.1* np.random.randn(*x.shape)
```

```python
y
```

```python
# Initialize Weights and Biases

w = np.random.randn()
b = np.random.randn()
```

```python
# Hyperparameters

lr= 0.1  #(lr = Learning rate)
epochs = 1000 # (epochs = Number of iterations)
```

```python
losses = [] # List to store loss values
```

```python
for epoch in range(epochs):

    y_pred = w * x + b
    
    # Compute loss (Mean Squared Error)
    loss = np.mean((y - y_pred) ** 1)
    
    # Store the loss value
    losses.append(loss)
    
    # (Gradient Descent)

    dw = np.mean(2*x * (y_pred-y))  # Gradient w.r.t. w
    db = np.mean(2*(y_pred-y))         # Gradient w.r.t. b
    
    # Update weights and biases
    w -= lr * dw          # w = w - learning_rate * gradient_w (lr*dw)
    b -= lr * db

    print(f'Epoch [{epoch + 1}/{epoch}], Loss: {loss:.4f}, w: {w:.4f}, b: {b:.4f}')   # 4f: Floating point with 4 decimal places
```

```python
plt.plot(range(1, epochs+1), losses)
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('Loss over Epochs')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(x, y, label='Data Points', color='blue')
plt.plot(x, w * x + b, label='Fitted Line', color='red',linewidth=2)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Linear Regression Fit')
plt.legend()
plt.grid(True)
plt.show()   
```
