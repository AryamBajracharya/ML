```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import svm

X = np.array([[1, 2], [2, 3], [3, 3], [6, 5], [7, 8], [8, 8]])
y = np.array([0, 0, 0, 1, 1, 1])

clf = svm.SVC(kernel='linear')
clf.fit(X, y)

# Plot data points
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='winter')

# Plot decision boundary (solid line)
ax = plt.gca()
xlim = ax.get_xlim()
xx = np.linspace(xlim[0], xlim[1], 100)
yy = -(clf.coef_[0][0] * xx + clf.intercept_) / clf.coef_[0][1]
plt.plot(xx, yy, 'k-')

plt.title("SVM Decision Boundary")
plt.show()
```

```python

clf = svm.SVC(kernel='linear')
clf.fit(X, y)

# Plot data points
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='winter')

# Plot support vectors (points defining the margin)
plt.scatter(clf.support_vectors_[:, 0], 
            clf.support_vectors_[:, 1], 
            s=100, facecolors='none', edgecolors='red')

# Calculate margin width (simplified)
margin = 1 / np.sqrt(np.sum(clf.coef_ ** 2))
print(f"Margin width: {margin:.2f}")

plt.title("SVM Margin (Red Circles = Support Vectors)")
plt.show()
```
