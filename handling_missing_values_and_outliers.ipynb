```python
#Box Plot
data = {
    "student": ['Aryam','Aavash','Nahakul','Bibek','Samyak','Sujan','Aryam'],
    "marks": [85,90,78,400,88,0,95]
}

df = pd.DataFrame(data)
df
```

```python
Q1 = df['marks'].quantile(0.25)
Q3 = df['marks'].quantile(0.75)
IQR = Q3 - Q1

print (Q1, Q3, IQR)

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier = df[(df['marks'] < 0) | (df['marks'] > 100)]  #Replace the value to 0 = lower_bound and 100 = upper_bound
```

```python
# Imputing Values
data = {
    "student": ['Aryam','Aavash','Nahakul','Bibek','Samyak','Sujan','Aryam'],
    "marks": [85,90,np.nan,np.nan,88,np.nan,95]
}

df = pd.DataFrame(data)
df
```

```python
df.fillna(df['marks'].mean()) #calculates mean and adds it on null values
```
